```python
import numpy as np
from pathlib import Path
import pandas as pd
from sklearn.metrics import log_loss

TRAINING_PATH = Path.cwd().parent / "training.csv.gz"
TESTING_PATH = Path.cwd().parent / "testing.csv.gz"

train = pd.read_csv(TRAINING_PATH)   # gzip читается напрямую, не распаковываем
test = pd.read_csv(TESTING_PATH )

print('Looking at Shape of TRAIN and TEST files')
print(train.shape, test.shape)
print('\nData Types of TRAIN file')
print(train.dtypes)
print('\nHead of TRAIN file')
print(train.head(3).T)                   # .T транспонирует таблицу, так удобнее читать много колонок
```

```python
prc_of_shots_made = 100 * (train['outcome'].value_counts(normalize=True))
print(prc_of_shots_made)
prc_of_shots_contested = 100 * (train['contested'].value_counts(normalize=True))
print(prc_of_shots_contested)
print('\nSum of nulls in Contesters')
contest1 = train['contester1_id'].isnull().sum()
contest2 = train['contester2_id'].isnull().sum()
contest3 = train['contester3_id'].isnull().sum()
contest4 = train['contester4_id'].isnull().sum()
print(contest1, contest2, contest3, contest4)
print('\nSum of nulls in DistCount')
dist1 = train['distcont1'].isnull().sum()
dist2 = train['distcont2'].isnull().sum()
dist3 = train['distcont3'].isnull().sum()
dist4 = train['distcont4'].isnull().sum()
print(dist1, dist2, dist3, dist4)

print('\nSum of 3P')
three_p_count = train['three'].sum()
print(three_p_count)

print('\nSum of all shots made')
total_shots = len(train['shot_id'])
print(total_shots)

print('\nHow many seasons are there?')
print(train['season_id'].value_counts())

print('Average shotclock (time - out of 24 seconds - at which shot was executed)')
print(train['shotclock'].describe())

print(test['season_id'].value_counts())

print('\nСколько нету в трэйне?')
```
